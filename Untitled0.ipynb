```python
!pip install yfinance -q

import yfinance as yf
import pandas as pd

# Download Reliance Industries - 10 years daily data
data = yf.download(
    "RELIANCE.NS",
    start="2016-01-01",
    end="2026-01-01",
    interval="1d",
    auto_adjust=False,
    progress=False
)

# Check the downloaded data
print("Number of observations:", len(data))
print("Starting date:", data.index.min())
print("Ending date:", data.index.max())

# Display first and last 5 observations
print("\nFirst 5 observations:")
print(data.head())

print("\nLast 5 observations:")
print(data.tail())

# Save to CSV
data.to_csv("Reliance_2016_2025.csv")

print("\nFile saved as: Reliance_2016_2025.csv")
```
